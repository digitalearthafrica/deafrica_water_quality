# Filter waterbodies by area

This notebook tests filtering the polygons in the Historical Extent product by area 
to determine the waterbodies to generate the per waterbody summary statistics. 

Latest version of the Waterbodies Historical Extent Product is 0.0.3 .

```python
# shapefile_url = "https://deafrica-services.s3.af-south-1.amazonaws.com/waterbodies/v0.0.3/historical_extent/waterbodies.shp.zip"
```

```python
waterbodies_s3_uri = "s3://deafrica-services/waterbodies/v0.0.3/historical_extent/waterbodies.shp.zip"
```

```python
import geopandas as gpd
```

```python
%%time
waterbodies_gdf = gpd.read_file(
    f"zip+{waterbodies_s3_uri}", 
    storage_options={"anon": True}
)
waterbodies_gdf.head()
```

```python
print(f"Loaded {len(waterbodies_gdf)} waterbody polygons")
```

```python
m2_per_ha = 10_000
area_threshold_ha = 0.1
```

```python
area_m2_ha = waterbodies_gdf["area_m2"] / m2_per_ha
area_filter = area_m2_ha > area_threshold_ha
```

```python
filtered_waterbodies = waterbodies_gdf[area_filter]
print(f"Number of waterbodies after area filtering to larger than {area_threshold_ha} ha: {len(filtered_waterbodies)} waterbody polygons")
```

```python
print(f"Number of polygons filtered out : {len(filtered_waterbodies) - len(waterbodies_gdf)}")
```

```python
print(f"Smallest waterbody area: {area_m2_ha.min()} ha")
```
